# ndim -

```python
# ndim - used to find dimension of array

import numpy as np
a=np.array([[10,20,30],[40,50,60]])
print(a.ndim)
```

```python

# shape -  used to find how many rows and column in a array

a=np.array([[10,20,30],[40,50,60]])
print(a.shape)
```

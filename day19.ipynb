```python
import matplotlib.pyplot as plt
```

# Line Chart

```python
x=[1,2,3,4,5]
y=[100, 150, 200, 250, 300]
plt.plot(x,y)
plt.title("Line Chart")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.show()
```

# Bar Chart

```python
x=[1,2,3,4,5]
y=["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
plt.bar(x,y, color="red")
plt.title("Bar Chart")
plt.show()
```

```python
x=[1,2,3,4,5]
y=["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
plt.barh(x,y, color="red")
plt.title("Bar Chart")
plt.show()
```

# Hist Chart

```python
a=[10,12,1,4,14,15,16,17,18,22,25]
plt.hist(a, bins=1)
plt.title("Hist Chart")
plt.show()
```

```python
a=[10,12,1,4,14,15,16,17,18,22,25]
plt.hist(a, bins=3)
plt.title("Hist Chart")
plt.show()
```

# Scatter Chart

```python
x=[1,2,3,4,5]
y=["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
plt.scatter(x,y, color="red")
plt.title("Scatter Chart")
plt.show()
```

```python
x=[5,6,3,4,2,10]
plt.pie(x)
plt.title("pie Chart")
plt.show()
```

# Box Plot

```python
a=[10,12,1,4,14,15,16,17,18,22,25, 100, 400, 200]
plt.boxplot(a)        #for finding outliers of your data
plt.show()
```
